**6. Create Bootstrap samples using Artificial Characters data set (number
of bootstrap samples = 10).
Report the cardinality of the training and testing sets in each sample.’**

```python
import numpy as np
from sklearn.datasets import fetch_openml
from sklearn.utils import resample
```

Load Data

```python
# Load Artificial Characters dataset (from OpenML)
data = fetch_openml(name="Artificial-Characters", version=1, as_frame=False)

X = data.data      # features
y = data.target    # labels

n_samples = X.shape[0]
print("Dataset shape:", X.shape)
```

Bootstrap

Each bootstrap sample:

* Training set: size = N (sampling with replacement)

* Testing set: Out-of-Bag samples (≈ 36.8% of N)

Theoretical Insight
For bootstrap sampling:

    𝑃(sample not selected)=(1−1/𝑁)^𝑁≈𝑒^−1≈0.368

So the test (OOB) set ≈ 36.8% of total samples.

```python
# Create 10 Bootstrap Samples
n_bootstrap = 10

for i in range(n_bootstrap):
    # Bootstrap sampling (with replacement)
    indices = resample(
        np.arange(n_samples),
        replace=True,
        n_samples=n_samples,
        random_state=i
    )

    X_train = X[indices]
    y_train = y[indices]

    # Out-of-Bag (test set)
    unique_train = np.unique(indices)
    test_indices = np.setdiff1d(np.arange(n_samples), unique_train)

    X_test = X[test_indices]
    y_test = y[test_indices]

    print(f"\nBootstrap Sample {i+1}")
    print("Training set cardinality:", len(X_train))
    print("Testing set cardinality :", len(X_test))
```
